## 4. Model Training

Train multiple machine learning models for comparison.

### 4.0 Baseline: Logistic Regression (TensorFlow)

We start with a simple logistic regression baseline using TensorFlow/Keras to establish a performance benchmark.

```python
# Train logistic regression baseline
lr_model, lr_history = models.train_logistic_regression_model(
    X_train=X_train, 
    Y_train=Y_train, 
    X_val=X_val, 
    Y_val=Y_val, 
    verbose=1
)

print("\n✓ Logistic Regression baseline training complete!")
```

```python
# Plot training history for logistic regression
plt.figure(figsize=(12, 4))

# Plot accuracy
plt.subplot(1, 2, 1)
plt.plot(lr_history.history['accuracy'], label='Training Accuracy')
plt.plot(lr_history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Logistic Regression - Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot loss
plt.subplot(1, 2, 2)
plt.plot(lr_history.history['loss'], label='Training Loss')
plt.plot(lr_history.history['val_loss'], label='Validation Loss')
plt.title('Logistic Regression - Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 4.1 Random Forest Model

```python
# Train Random Forest with hyperparameter tuning
rf_model = models.train_random_forest_model(
    X_train=X_train, Y_train=Y_train, X_val=X_val, Y_val=Y_val, verbose=2
)

print("\n✓ Random Forest training complete!")
```

### 4.2 XGBoost Model

```python
# Train XGBoost with hyperparameter tuning
xgb_model = models.train_xgboost_model(
    X_train=X_train, Y_train=Y_train, X_val=X_val, Y_val=Y_val, verbose=0
)

print("\n✓ XGBoost training complete!")
```

### 4.3 Transformer Model 

Train a neural network-based transformer model.

```python
# Train Transformer model
transformer_model, history = models.train_transformer_model(
    X_train=X_train,
    Y_train=Y_train,
    X_val=X_val,
    Y_val=Y_val,
    epochs=10,
    batch_size=64,
    verbose=1,
)

print("\n✓ Transformer training complete!")
```

```python
# Plot training history
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Plot accuracy
ax1.plot(history.history["accuracy"], label="Train Accuracy")
ax1.plot(history.history["val_accuracy"], label="Val Accuracy")
ax1.set_title("Model Accuracy")
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Accuracy")
ax1.legend()
ax1.grid(alpha=0.3)

# Plot loss
ax2.plot(history.history["loss"], label="Train Loss")
ax2.plot(history.history["val_loss"], label="Val Loss")
ax2.set_title("Model Loss")
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Loss")
ax2.legend()
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### 5.4 Model Comparison

```python
# Compare all models
models_dict = {
    "Logistic Regression": lr_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model,
    "Transformer": transformer_model,
}

comparison_df = models.compare_models(
    models_dict=models_dict,
    X_train=X_train,
    Y_train=Y_train,
    X_val=X_val,
    Y_val=Y_val,
    X_test=X_test,
    Y_test=Y_test,
)

print("\n" + "=" * 70)
print("MODEL COMPARISON SUMMARY")
print("=" * 70)
display(comparison_df)
```

```python
# Visualize model comparison
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

metrics_to_plot = ['Accuracy', 'Precision', 'Recall', 'F1 Score']
datasets = ['Train', 'Val', 'Test']

for idx, metric in enumerate(metrics_to_plot):
    row, col = idx // 2, idx % 2
    ax = axes[row, col]
    
    # Prepare data for plotting
    plot_data = []
    for dataset in datasets:
        col_name = f'{dataset} {metric}'
        if col_name in comparison_df.columns:
            plot_data.append(comparison_df[col_name].values)
    
    x = np.arange(len(comparison_df))
    width = 0.25
    
    for i, dataset in enumerate(datasets):
        col_name = f'{dataset} {metric}'
        if col_name in comparison_df.columns:
            ax.bar(x + i*width, comparison_df[col_name], width, label=dataset)
    
    ax.set_xlabel('Model')
    ax.set_ylabel(metric)
    ax.set_title(f'{metric} Comparison')
    ax.set_xticks(x + width)
    ax.set_xticklabels(comparison_df['Model'], rotation=45, ha='right')
    ax.legend()
    ax.grid(axis='y', alpha=0.3)
    ax.set_ylim([0, 1])

plt.tight_layout()
plt.show()
```
